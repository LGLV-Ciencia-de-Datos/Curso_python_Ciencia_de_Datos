## **Selección de Datos para el Modelado**

Tu conjunto de datos tenía demasiadas variables para asimilarlas, o incluso para imprimirlas de forma ordenada. ¿Cómo puedes reducir esta abrumadora cantidad de datos a algo que puedas entender?

Comenzaremos por elegir algunas variables basándonos en nuestra intuición. Cursos posteriores te mostrarán técnicas estadísticas para priorizar variables automáticamente.

Para elegir variables/columnas, necesitaremos ver una lista de todas las columnas en el conjunto de datos. Eso se hace con la propiedad columns del DataFrame (la última línea de código a continuación).

```python
import pandas as pd
```

```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("dansbecker/melbourne-housing-snapshot")

print("Path to dataset files:", path)

# Load the data into a pandas DataFrame
melbourne_file_path = path + '/melb_data.csv'
melbourne_data = pd.read_csv(melbourne_file_path)

# Print the first 5 rows of the DataFrame
display(melbourne_data)
```

```python
len(melbourne_data.columns)
```

```python
melbourne_data.isna().sum()
```

```python
# El conjunto de datos de Melbourne tiene algunos valores faltantes
 # (algunas casas para las que no se registraron ciertas variables).

# Aprenderemos a manejar los valores faltantes en un tutorial posterior.
# El conjunto de datos de Iowa no tiene valores faltantes en las columnas que usarás.
# Así que, por ahora, tomaremos la opción más simple y eliminaremos las casas de nuestros datos.
# No te preocupes mucho por esto ahora, aunque el código es:

# dropna elimina los valores faltantes (piensa en na como "no disponible")

melbourne_data = melbourne_data.dropna(axis=0)
```

```python
melbourne_data
```

Hay muchas maneras de seleccionar un subconjunto de tus datos. El curso de Pandas cubre estas opciones con mayor profundidad, pero por ahora nos centraremos en dos enfoques.

* **Notación de punto**, que usamos para seleccionar el "objetivo de predicción".

* **Selección con una lista de columnas**, que usamos para seleccionar las "características".

## **Seleccionar la Variable a Predecir**

Puedes extraer una variable usando la notación de punto. Esta columna única se almacena en una `Series`, que es muy similar a un `DataFrame` pero con una sola columna de datos.

Usaremos la notación de punto para seleccionar la columna que queremos predecir, la cual se llama "variable a predecir" (o `prediction target`). Por convención, a la variable a predecir se le llama `y`. Entonces, el código que necesitamos para guardar los precios de las casas en los datos de Melbourne es:

```python
y = melbourne_data.Price # y es la variable dependiente,  variable a predecir o target.
```

## **Seleccionando "Features" (Características)**

Las columnas que se introducen en nuestro modelo (y que luego se usan para hacer predicciones) se llaman "features" (características). En nuestro caso, serían las columnas utilizadas para determinar el precio de la casa. A veces, usarás todas las columnas excepto la de destino (`target`) como características. Otras veces, te irá mejor con menos características.

Por ahora, construiremos un modelo con solo unas pocas características. Más adelante verás cómo iterar y comparar modelos construidos con diferentes características.

Seleccionamos múltiples características proporcionando una lista de nombres de columnas dentro de corchetes. Cada elemento de esa lista debe ser una cadena de texto (entre comillas).

Aquí tienes un ejemplo:

```python
melbourne_data.columns
```

```python
melbourne_features = ['Rooms', 'Bathroom', 'Landsize', 'Lattitude', 'Longtitude'] # aca se seleccionana las columnas o variables independientes.
```

Por convención, a estos datos se les llama X.

```python
X = melbourne_data[melbourne_features]
```

Revisemos rápidamente los datos que usaremos para predecir los precios de las casas usando el método `describe` y el método `head`, que muestra las primeras filas.

```python
X.describe()
```

```python
X.head()
```

Inspeccionar tus datos visualmente con estos comandos es una parte importante del trabajo de un científico de datos. Con frecuencia encontrarás sorpresas en el conjunto de datos que merecen una inspección más detallada.

## **Construyendo tu modelo**

Utilizarás la librería `scikit-learn` para crear tus modelos. Cuando se escribe código, esta librería se llama `sklearn`, como verás en el código de ejemplo. `Scikit-learn` es, sin duda, la librería más popular para modelar los tipos de datos que se suelen almacenar en los `DataFrames`.

Los pasos para construir y usar un modelo son:

* Definir: ¿Qué tipo de modelo será? ¿Un árbol de decisión? ¿Algún otro tipo de modelo? También se especifican otros parámetros del tipo de modelo.

* Ajustar (`Fit`): Captura patrones de los datos proporcionados. Este es el corazón del modelado.

* Predecir: Justo como suena.

* Evaluar: Determina qué tan precisas son las predicciones del modelo.

Aquí tienes un ejemplo de cómo definir un modelo de árbol de decisión con `scikit-learn` y ajustarlo con las variables de características y la variable objetivo.

```python
from sklearn.tree import DecisionTreeRegressor
```

```python
# Define el modelo. Especifica un número de random_state que asegura el mismo resultado en cada corrida.
melbourne_model = DecisionTreeRegressor(random_state=1)

# Fit model
melbourne_model.fit(X, y)
```

Muchos modelos de aprendizaje automático permiten cierta aleatoriedad en el entrenamiento. Especificar un número para `random_state` asegura que obtengas los mismos resultados en cada ejecución. Esto se considera una buena práctica. Puedes usar cualquier número, y la calidad del modelo no dependerá significativamente del valor exacto que elijas.

Ahora tenemos un modelo ajustado que podemos usar para hacer predicciones.

En la práctica, querrás hacer predicciones para casas nuevas que salgan al mercado en lugar de las casas de las que ya tenemos precios. Pero haremos predicciones para las primeras filas de los datos de entrenamiento para ver cómo funciona la función `predict`.

```python
print("Haciendo predicciones para las siguientes 5 casas:\n")

print(X.head())
print('\n')
print(f"Las predicciones son:{melbourne_model.predict(X.head())}\n")

print(f'El valor real de las casas es:\n{y.head()}')
```
