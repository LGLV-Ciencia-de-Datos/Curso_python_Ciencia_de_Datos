# **Introducción al Aprendizaje Automático** (Intro to Machine Learning).
Aprende las ideas fundamentales del aprendizaje automático y construye tus primeros modelos.

Comenzaremos con una visión general de cómo funcionan y se utilizan los modelos de aprendizaje automático (machine learning). Esto podría parecerte muy básico si ya tienes experiencia en modelos estadísticos o aprendizaje automático. No te preocupes, pronto avanzaremos para construir modelos mucho más potentes.

En este curso, construirás modelos siguiendo un escenario específico:

Tu primo ha ganado millones de dólares especulando con bienes raíces. Debido a tu interés en la ciencia de datos, te ha propuesto ser su socio. Él aportará el capital, y tú le proporcionarás modelos que predigan el valor de diferentes casas.

Le preguntas a tu primo cómo ha predicho los precios de las casas en el pasado, y él te dice que todo es cuestión de intuición. Sin embargo, al indagar más, descubres que ha identificado patrones de precios en casas que ha visto anteriormente y utiliza esos patrones para hacer predicciones sobre nuevas casas que está considerando comprar.

El aprendizaje automático funciona exactamente de la misma manera. Empezaremos con un modelo llamado Árbol de Decisión (Decision Tree). Existen modelos más sofisticados que ofrecen predicciones más precisas, pero los árboles de decisión son fáciles de entender y son el componente fundamental de algunos de los mejores modelos en la ciencia de datos.

Para empezar de forma sencilla, comenzaremos con el árbol de decisión más simple posible.
