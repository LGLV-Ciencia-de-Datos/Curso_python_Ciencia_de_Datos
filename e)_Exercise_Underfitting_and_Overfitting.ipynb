## Recapitulando

Has construido tu primer modelo y ahora es el momento de optimizar el tamaño del árbol para hacer mejores predicciones. Ejecuta estas celdas para configurar tu entorno de programación donde lo dejaste en el paso anterior.

```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("nickptaylor/iowa-house-prices")

print("Path to dataset files:", path)

import pandas as pd
import os

# Load the data into a pandas DataFrame
home_data_path = path + '/train.csv'
home_data = pd.read_csv(home_data_path)

# Print the first 5 rows of the DataFrame
display(home_data.head())

# Code you have previously used to load data

from sklearn.metrics import mean_absolute_error
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeRegressor

# Create target object and call it y
y = home_data.SalePrice
# Create X
features = ['LotArea', 'YearBuilt', '1stFlrSF', '2ndFlrSF', 'FullBath', 'BedroomAbvGr', 'TotRmsAbvGrd']
X = home_data[features]

# Split into validation and training data
train_X, val_X, train_y, val_y = train_test_split(X, y, random_state=1)

# Specify Model
iowa_model = DecisionTreeRegressor(random_state=1)
# Fit Model
iowa_model.fit(train_X, train_y)

# Make validation predictions and calculate mean absolute error
val_predictions = iowa_model.predict(val_X)
val_mae = mean_absolute_error(val_predictions, val_y)
print("Validation MAE: {:,.0f}".format(val_mae))
```
