# **Introducción**

¡Bienvenido al curso de Aprendizaje Automático Intermedio!
Si tienes experiencia en aprendizaje automático y quieres aprender a mejorar rápidamente la calidad de tus modelos, ¡estás en el lugar adecuado! En este curso, acelerarás tu conocimiento en aprendizaje automático aprendiendo a:

* Abordar tipos de datos que se encuentran a menudo en conjuntos de datos del mundo real con: valores faltantes, variables categóricas (`missing values`, `categorical variables`).
* Diseñar pipelines para mejorar la calidad de tu código de aprendizaje automático.
* Utilizar técnicas avanzadas de validación de modelos validación cruzada (`cross-validation`).
* Construir modelos de vanguardia que se utilizan ampliamente como (`XGBoost`). y
* Evitar errores comunes e importantes en ciencia de datos (fugas de información).

A lo largo del curso, aplicarás tus conocimientos completando ejercicios prácticos con datos del mundo real en cada nuevo tema. Los ejercicios prácticos utilizan datos de Precios de Viviendas, donde emplearás 79 variables explicativas diferentes (como el tipo de tejado, número de dormitorios y número de baños) para predecir los precios de las viviendas.
