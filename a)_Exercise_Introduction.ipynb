```python
import os

import kagglehub

# Download latest version
path = kagglehub.dataset_download("nickptaylor/iowa-house-prices")

print("Path to dataset files:", path)
```

```python
train_csv_path = path + '/train.csv'
test_csv_path = path + '/test.csv'

X_full = pd.read_csv(train_csv_path, index_col='Id')
X_test_full = pd.read_csv(test_csv_path, index_col='Id')
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split

# Use the already loaded dataframes
# X_full = pd.read_csv('../input/train.csv', index_col='Id') # Already loaded
# X_test_full = pd.read_csv('../input/test.csv', index_col='Id') # Already loaded

# Obtain target and predictors
y = X_full.SalePrice
features = ['LotArea', 'YearBuilt', '1stFlrSF', '2ndFlrSF', 'FullBath', 'BedroomAbvGr', 'TotRmsAbvGrd']
X = X_full[features].copy()
X_test = X_test_full[features].copy()

# Break off validation set from training data
X_train, X_valid, y_train, y_valid = train_test_split(X, y, train_size=0.8, test_size=0.2,
                                                      random_state=0)

print("Data loaded and split successfully!")
```

La siguiente celda de código define cinco modelos diferentes de bosques aleatorios. Ejecuta esta celda sin cambios.

```python
from sklearn.ensemble import RandomForestRegressor

# Define the models
model_1 = RandomForestRegressor(n_estimators=50, random_state=0)
model_2 = RandomForestRegressor(n_estimators=100, random_state=0)
model_3 = RandomForestRegressor(n_estimators=100, criterion='absolute_error', random_state=0)
model_4 = RandomForestRegressor(n_estimators=200, min_samples_split=20, random_state=0)
model_5 = RandomForestRegressor(n_estimators=100, max_depth=7, random_state=0)

models = [model_1, model_2, model_3, model_4, model_5]
```

Para seleccionar el mejor modelo de los cinco, definimos la función `score_model()` a continuación. Esta función devuelve el error absoluto medio `MAE` del conjunto de validación. Recordemos que el mejor modelo será aquel que obtenga el menor MAE.

```python
from sklearn.metrics import mean_absolute_error

# Define una función llamada score_model que toma un modelo (model)
# y conjuntos de datos de entrenamiento y validación como argumentos.
# Los argumentos X_t, X_v, y_t, y_v tienen valores predeterminados.
def score_model(model, X_t=X_train, X_v=X_valid, y_t=y_train, y_v=y_valid):

    # Entrena el modelo usando los datos de entrenamiento (X_t, y_t).
    # Esto ajusta los parámetros del modelo a los datos proporcionados.
    model.fit(X_t, y_t)

    # Usa el modelo ya entrenado para hacer predicciones (preds) sobre los
    # datos de validación (X_v).
    preds = model.predict(X_v)

    # Calcula y retorna el Error Absoluto Medio (MAE)
    # comparando las etiquetas reales (y_v) con las predicciones (preds).
    return mean_absolute_error(y_v, preds)

# ---

# Inicia un bucle for que itera a través de los modelos en la lista 'models'.
# La variable 'i' toma valores desde 0 hasta la cantidad de modelos menos 1.
for i in range(0, len(models)):

    # Llama a la función score_model para cada modelo en la lista.
    # El resultado, que es el MAE, se almacena en la variable 'mae'.
    mae = score_model(models[i])

    # Imprime una cadena formateada que muestra el número del modelo (i+1)
    # y su MAE correspondiente.
    print("Model %d MAE: %d" % (i+1, mae))
```

```python
# Find the index of the model with the minimum MAE
min_mae = float('inf')
best_model_index = -1

for i in range(len(models)):
    mae = score_model(models[i], X_train, X_valid, y_train, y_valid) # Recalculate MAE or use stored values if available
    if mae < min_mae:
        min_mae = mae
        best_model_index = i

# Extract the best model
best_model = models[best_model_index]

#print(f"The best model is Model {best_model_index + 1} with MAE: {(min_mae)}")
```
